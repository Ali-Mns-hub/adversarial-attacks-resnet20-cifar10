### Accuracy Before Attacks

```python
import torch
from torchvision import datasets, transforms
from torch.utils.data import DataLoader

# 1. Device setup – GPU if you got it, otherwise CPU
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# 2. Load ResNet-20 pretrained on CIFAR-10 via Torch Hub :contentReference[oaicite:0]{index=0}
model = torch.hub.load(
    "chenyaofo/pytorch-cifar-models",
    "cifar10_resnet20",
    pretrained=True
)
model.to(device)
model.eval()

# 3. Define transforms with your mean/std
mean = (0.4914, 0.4822, 0.4465)
std  = (0.2023, 0.1994, 0.2010)
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize(mean, std)
])

# 4. Prepare CIFAR-10 test set
testset = datasets.CIFAR10(
    root="./data",
    train=False,
    download=True,
    transform=transform
)
testloader = DataLoader(
    testset,
    batch_size=128,
    shuffle=False,
    num_workers=2
)

# 5. Class names as provided
classes = [
    'airplane', 'automobile', 'bird', 'cat', 'deer',
    'dog', 'frog', 'horse', 'ship', 'truck'
]

# 6. Evaluation loop
correct = 0
total = 0

# For per-class stats
class_correct = [0] * 10
class_total   = [0] * 10

with torch.no_grad():
    for images, labels in testloader:
        images, labels = images.to(device), labels.to(device)
        outputs = model(images)
        _, preds = outputs.max(1)
        correct += (preds == labels).sum().item()
        total += labels.size(0)
        # per-class
        for lbl, pred in zip(labels, preds):
            class_total[lbl] += 1
            if pred == lbl:
                class_correct[lbl] += 1

# 7. Print overall accuracy
overall_acc = 100 * correct / total
print(f" Overall CIFAR-10 accuracy: {overall_acc:.2f}% ")

# 8. Print per-class accuracy
for i, cls in enumerate(classes):
    acc = 100 * class_correct[i] / class_total[i]
    print(f" - {cls:10s}: {acc:5.2f}%")
```

### Part 1: FAST GRADIENT SIGN METHOD(FGSM)

```python
import torch
import torchvision
import torchvision.transforms as transforms
import torch.nn as nn
import matplotlib.pyplot as plt
import numpy as np
import torchattacks

# Load the pre-trained ResNet-20 model
model = torch.hub.load("chenyaofo/pytorch-cifar-models", "cifar10_resnet20", pretrained=True)

# Move model to the appropriate device (GPU or CPU)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)
model.eval()

# CIFAR-10 dataset transformation
mean = (0.4914, 0.4822, 0.4465)
std = (0.2023, 0.1994, 0.2010)
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize(mean, std),
])

# Load CIFAR-10 dataset
testset = torchvision.datasets.CIFAR10(root='./data', train=False, download=True, transform=transform)
testloader = torch.utils.data.DataLoader(testset, batch_size=64, shuffle=False)

# Initialize the FGSM attack using the torchattacks library
attack = torchattacks.FGSM(model, eps=0.1)

# CIFAR-10 class labels
classes = ['airplane', 'automobile', 'bird', 'cat', 'deer', 'dog', 'frog', 'horse', 'ship', 'truck']

# Function to display images with class labels
def display_images(original, noise, perturbed, label, pred, idx):
    # Convert the tensor to numpy array and unnormalize the image
    original = original.squeeze().cpu().detach().numpy().transpose((1, 2, 0)) * np.array(std) + np.array(mean)
    noise = noise.squeeze().cpu().detach().numpy().transpose((1, 2, 0))
    perturbed = perturbed.squeeze().cpu().detach().numpy().transpose((1, 2, 0)) * np.array(std) + np.array(mean)

    # Plot the original, noise, and perturbed images
    plt.figure(figsize=(12, 4))

    # Original image
    plt.subplot(1, 3, 1)
    plt.imshow(original)
    plt.title(f"Original: {classes[label]} ({label})")
    plt.axis('off')

    # Noise (perturbation)
    plt.subplot(1, 3, 2)
    plt.imshow(noise)
    plt.title(f"Noise")
    plt.axis('off')

    # Perturbed image
    plt.subplot(1, 3, 3)
    plt.imshow(perturbed)
    plt.title(f"Adversarial: {classes[pred]} ({pred})")
    plt.axis('off')

    plt.show()

# Evaluate the model on adversarial examples
incorrect_images = []
correct = 0
total = 0

for data, target in testloader:
    data, target = data.to(device), target.to(device)

    # Generate adversarial images using FGSM from the torchattacks library
    perturbed_data = attack(data, target)

    # Re-classify the perturbed image
    output = model(perturbed_data)
    _, predicted = torch.max(output, 1)
    total += target.size(0)
    correct += (predicted == target).sum().item()

    # Collect images where the model misclassified the adversarial example
    misclassified_idx = (predicted != target).cpu().numpy()
    for i in range(len(misclassified_idx)):
        if misclassified_idx[i]:
            if len(incorrect_images) < 5:  # Stop after collecting 5 images
                noise = perturbed_data[i] - data[i]  # Calculate the noise added
                incorrect_images.append((data[i], perturbed_data[i], noise, target[i].item(), predicted[i].item()))

# Print out the final accuracy on the adversarial examples
print(f'Accuracy on adversarial examples: {100 * correct / total}%')

# Display 5 incorrect adversarial examples
for idx, (original, perturbed, noise, label, pred) in enumerate(incorrect_images):
    display_images(original, noise, perturbed, label, pred, idx)
```

### Part 3: PGD WITH MOMENTUM

```python
import torch
import torchvision
import torchvision.transforms as transforms
import torch.nn as nn
import matplotlib.pyplot as plt
import numpy as np

# Load the pre-trained ResNet-20 model
model = torch.hub.load("chenyaofo/pytorch-cifar-models", "cifar10_resnet20", pretrained=True)

# Move model to the appropriate device (GPU or CPU)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)
model.eval()

# CIFAR-10 dataset transformation
mean = (0.4914, 0.4822, 0.4465)
std = (0.2023, 0.1994, 0.2010)
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize(mean, std),
])

# Load CIFAR-10 dataset
testset = torchvision.datasets.CIFAR10(root='./data', train=False, download=True, transform=transform)
testloader = torch.utils.data.DataLoader(testset, batch_size=64, shuffle=False)

class PGDWithMomentum:
    def __init__(self, model, eps, alpha, steps, momentum=1.0):
        self.model = model
        self.eps = eps  # Maximum perturbation
        self.alpha = alpha  # Step size
        self.steps = steps  # Number of iterations
        self.momentum = momentum  # Momentum factor

    def __call__(self, images, labels):
        images = images.clone().detach().to(device)
        labels = labels.clone().detach().to(device)

        perturbed_images = images.clone().detach()
        perturbed_images.requires_grad = True

        velocity = torch.zeros_like(perturbed_images).to(device)

        for _ in range(self.steps):
            outputs = self.model(perturbed_images)
            loss = nn.CrossEntropyLoss()(outputs, labels)

            self.model.zero_grad()
            loss.backward()

            # Get the normalized gradient
            grad = perturbed_images.grad.detach()

            velocity = self.momentum * velocity + grad / (grad.abs().mean() + 1e-8)  # small epsilon to avoid division by 0

            # Update the adversarial image
            perturbed_images = perturbed_images.detach() + self.alpha * velocity.sign()

            # Project back into the epsilon-ball and valid pixel range
            perturbed_images = torch.max(torch.min(perturbed_images, images + self.eps), images - self.eps)
            perturbed_images = torch.clamp(perturbed_images, 0, 1)

            # Make sure perturbed_images can track gradients again
            perturbed_images.requires_grad = True

        return perturbed_images


# Initialize the custom PGD with Momentum attack
attack = PGDWithMomentum(model, eps=0.1, alpha=0.01, steps=40, momentum=1.0)

# CIFAR-10 class labels
classes = ['airplane', 'automobile', 'bird', 'cat', 'deer', 'dog', 'frog', 'horse', 'ship', 'truck']

# Function to display images with class labels
def display_images(original, noise, perturbed, label, pred, idx):
    # Convert the tensor to numpy array and unnormalize the image
    original = original.squeeze().cpu().detach().numpy().transpose((1, 2, 0)) * np.array(std) + np.array(mean)
    noise = noise.squeeze().cpu().detach().numpy().transpose((1, 2, 0))
    perturbed = perturbed.squeeze().cpu().detach().numpy().transpose((1, 2, 0)) * np.array(std) + np.array(mean)

    # Plot the original, noise, and perturbed images
    plt.figure(figsize=(12, 4))

    # Original image
    plt.subplot(1, 3, 1)
    plt.imshow(original)
    plt.title(f"Original: {classes[label]} ({label})")
    plt.axis('off')

    # Noise (perturbation)
    plt.subplot(1, 3, 2)
    plt.imshow(noise)
    plt.title(f"Noise")
    plt.axis('off')

    # Perturbed image
    plt.subplot(1, 3, 3)
    plt.imshow(perturbed)
    plt.title(f"Adversarial: {classes[pred]} ({pred})")
    plt.axis('off')

    plt.show()

# Evaluate the model on adversarial examples
incorrect_images = []
correct = 0
total = 0

for data, target in testloader:
    data, target = data.to(device), target.to(device)

    # Generate adversarial images using custom PGD with Momentum
    perturbed_data = attack(data, target)

    # Re-classify the perturbed image
    output = model(perturbed_data)
    _, predicted = torch.max(output, 1)
    total += target.size(0)
    correct += (predicted == target).sum().item()

    # Collect images where the model misclassified the adversarial example
    misclassified_idx = (predicted != target).cpu().numpy()
    for i in range(len(misclassified_idx)):
        if misclassified_idx[i]:
            if len(incorrect_images) < 5:  # Stop after collecting 5 images
                noise = perturbed_data[i] - data[i]  # Calculate the noise added
                incorrect_images.append((data[i], perturbed_data[i], noise, target[i].item(), predicted[i].item()))

# Print out the final accuracy on the adversarial examples
print(f'Accuracy on adversarial examples: {100 * correct / total}%')

# Display 5 incorrect adversarial examples
for idx, (original, perturbed, noise, label, pred) in enumerate(incorrect_images):
    display_images(original, noise, perturbed, label, pred, idx)
```

### Part 4:AUTOPGD WITH CE & CW LOSSES

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

def cw_loss(logits: torch.Tensor, labels: torch.Tensor) -> torch.Tensor:
    """
    Untargeted CW loss: f(x,y) = max_{i != y} z_i - z_y
    Returns: tensor of shape [batch]
    """
    # logits: [batch, C], labels: [batch]
    one_hot = F.one_hot(labels, num_classes=logits.size(1)).bool()
    real = logits[one_hot].view(logits.size(0), 1)               # [batch,1]
    other = logits.masked_fill(one_hot, float('-inf'))         # [batch,C]
    max_other, _ = other.max(dim=1, keepdim=True)              # [batch,1]
    return (max_other - real).squeeze(1)                        # [batch]

def auto_pgd(
    model: nn.Module,
    x_orig: torch.Tensor,
    y: torch.Tensor,
    eps: float,
    n_iter: int,
    loss_fn: str = 'ce',       # 'ce' or 'cw'
    alpha: float = 0.75,
    rho: float = 0.75,
) -> torch.Tensor:
    """
    Auto-PGD attack (l_inf) on a single batch x_orig, y.
    loss_fn: 'ce' for cross-entropy, 'cw' for Carlini-Wagner.
    Returns: adversarial batch of same shape as x_orig.
    """
    model.eval()
    batch_size = x_orig.size(0)
    device = x_orig.device

    # Prepare loss
    ce_loss = nn.CrossEntropyLoss(reduction='none')
    def f(x):
        logits = model(x)
        if loss_fn == 'ce':
            return ce_loss(logits, y)
        else:
            return cw_loss(logits, y)

    # compute checkpoint schedule w_j
    p = [0.0, 0.22]
    while p[-1] < 1.0:
        pj, pj_1 = p[-1], p[-2]
        step = max(pj - pj_1 - 0.03, 0.06)
        p.append(min(pj + step, 1.0))
    w = sorted(set(int(pi * n_iter) for pi in p))
    # ensure 0 and n_iter included
    if w[0] != 0: w.insert(0, 0)
    if w[-1] != n_iter: w.append(n_iter)

    # Initialization
    x_prev = x_orig.clone().detach()
    x = torch.clamp(x_orig + torch.empty_like(x_orig).uniform_(-eps, eps), 0, 1)
    x = x.detach().requires_grad_(True)
    f_prev = f(x_prev)       # [batch]
    f_curr = f(x)            # [batch]
    fmax, idx_max = torch.max(torch.stack([f_prev, f_curr], dim=0), dim=0)
    x_max = torch.where(
        f_curr.unsqueeze(1).unsqueeze(2).unsqueeze(3) >= f_prev.unsqueeze(1).unsqueeze(2).unsqueeze(3),
        x, x_prev
    )

    eta = 2 * eps              # initial step‐size
    last_halved = False
    succ_count = torch.zeros(batch_size, device=device)

    for k in range(1, n_iter):
        # 1) Pure gradient step
        x.grad = None
        logits = model(x)
        loss = ce_loss(logits, y).sum() if loss_fn=='ce' else cw_loss(logits, y).sum()
        loss.backward()
        grad = x.grad.data
        z = x + eta * grad.sign()
        # project into eps-ball and [0,1]
        z = torch.max(torch.min(z, x_orig + eps), x_orig - eps).clamp(0,1)

        # 2) Momentum update
        x_new = x + alpha*(z - x) + (1-alpha)*(x - x_prev)
        x_prev = x.detach()
        x = torch.max(torch.min(x_new, x_orig + eps), x_orig - eps).clamp(0,1)
        x = x.detach().requires_grad_(True)

        # 3) Update f, track best
        f_curr = f(x)
        improved = (f_curr > fmax)
        fmax = torch.where(improved, f_curr, fmax)
        x_max = torch.where(
            improved.view(-1,1,1,1), x.detach(), x_max
        )

        # count successes for checkpoint
        succ_count += (f_curr > f_prev).float()
        f_prev = f_curr.clone()

        # 4) Checkpoint logic
        if k in w[1:-1]:
            j = w.index(k)
            period = w[j] - w[j-1]
            cond1 = succ_count < rho * period
            cond2 = (~last_halved) & (fmax == f_prev.min())
            if cond1.any() or cond2.any():
                eta = eta / 2
                x = x_max.detach().requires_grad_(True)
                last_halved = True
            else:
                last_halved = False
            succ_count.zero_()

    return x_max.detach()

# Convenience wrappers

def auto_pgd_ce(model, x, y, eps, n_iter=100):
    return auto_pgd(model, x, y, eps, n_iter, loss_fn='ce')

def auto_pgd_cw(model, x, y, eps, n_iter=100):
    return auto_pgd(model, x, y, eps, n_iter, loss_fn='cw')
```
